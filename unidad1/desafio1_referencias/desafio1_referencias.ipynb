# Desafío 1: El misterio de las referencias

## 1. Objetivo

Comprender cómo funcionan las referencias a objetos en Python y analizar qué sucede cuando dos variables hacen referencia a la misma lista.

## 2. Planteamiento del problema

Un estudiante intenta duplicar una lista de valores, pero al modificar la supuesta copia, la lista original también cambia.

El objetivo es identificar por qué sucede este comportamiento y corregir el código para que ambas listas sean independientes.

```python
original_data = [10, 20, 30]

# El estudiante intenta copiar los datos
copy_data = original_data

copy_data.append(40)

print(f"Original: {original_data}")
print(f"Copia: {copy_data}")
```

## 3. Análisis del error

El problema se encuentra en la siguiente instrucción:

`copy_data = original_data`

Esta instrucción no crea una nueva lista. En Python, `copy_data` y `original_data` hacen referencia al mismo objeto en memoria.

Por esta razón, cuando se ejecuta `copy_data.append(40)`, se modifica la misma lista a la que hace referencia `original_data`.

Conceptualmente:

original_data ──┐
                ├──> [10, 20, 30]
copy_data ──────┘

Después de agregar el número 40:

original_data ──┐
                ├──> [10, 20, 30, 40]
copy_data ──────┘

## 4. Solución

Para crear una lista independiente se utilizará el método `copy()`.

De esta manera, se crea un nuevo objeto con los mismos elementos de la lista original, evitando que las modificaciones realizadas sobre la copia afecten a la lista original.

```python
original_data = [10, 20, 30]

# Crear una copia independiente
copy_data = original_data.copy()

# Modificar solamente la copia
copy_data.append(40)

print(f"Original: {original_data}")
print(f"Copia: {copy_data}")
```

## 5. Comprobación

Para comprobar que ambas variables representan objetos diferentes, se puede utilizar el operador `is`.


```python
print("¿Son el mismo objeto?", original_data is copy_data)
```

## 6. Conclusión

El ejercicio permitió comprender la diferencia entre asignar una referencia y crear una copia independiente.

Cuando se utiliza `copy_data = original_data`, ambas variables hacen referencia al mismo objeto. En cambio, mediante `original_data.copy()` se crea una nueva lista independiente.

Este concepto es importante para comprender cómo Python maneja los objetos y las referencias en memoria.
